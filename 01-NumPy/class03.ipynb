```python
import numpy as np
```

# NumPy array:

```python
x = np.array([1,2,3,4,5])
print(x)
print(type(x))

```

```python
# np.zeros(), np.ones(), np.full()
print(np.zeros(5)) # by Default Float values 
print(np.zeros(5,dtype='int32')) # int values
print(np.zeros((3,4),dtype='int')) # 2D array formation

print(np.ones((2,4,5)))
```

```python
# np.arange(start , stop , step) , stop is excluded.
print(np.arange(10,5,-2))
print(np.arange(10))
print(np.arange(1,11).reshape(2,5)) # reshape should be used in a way that number of elements are equal in arange and reshape
```

```python
# np.linspace(start , stop , num)  it divides the range from start to end in num equal parts
print(np.linspace(1,10,3))
print(np.linspace(1,10,2))
print(np.linspace(1,10,5))
```

```python
# np.eye(n) , gives a identity matrix of n*n 
# print(np.eye(4))
# print(np.eye(3,4))
print(np.eye(3,4,k=2)) # k=2 shifts the principle diagonal 
```

### Array Attributes


```python
# .shape .ndim .size .dtype
x = np.arange(20).reshape(4,5)
print(x)
print(x.shape)

# ndim tells number of dimensions
# size tells number of elements

print(len(x)) # tells the length of the 1st dimesion only

# dtype tells the datatype
print(type(x))
print(x.dtype)
```

### Indexing and Slicing

```python
arr = np.arange(10)
print(arr)
print(arr[-3]) # negative indexing
print(arr[-3:-12:-2])
```

```python
# 2D indexing - arr[row ,col] , full row , full column
arr = np.arange(20).reshape(5,4)
print(arr)
print(arr[2][2]) # cell index
```

```python
# Subarray Slicing
# print(arr)
# print(arr[2: , 1:])
# print(arr[::4 , ::3]) # print all the corner elements.
print(arr[1:3:1 , -2::-1])
```
