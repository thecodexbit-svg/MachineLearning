```python
import numpy as np
```

```python
print(np.array([1,2.5,"hello"])) # it should throw error but it converts every elment to the string type
x = np.array([1,4,4.5,9.8], dtype='int') # we can also define the data type but decimal value is removed not rounded off.
print(x)
```

```python
y = np.array([1,4,4.5,9.8])
print(y.astype('int')) # .astype creates a new copy and do not change the original array.
```

```python
# arithmetic - array with scalar, array with array
arr = np.arange(10)
print(arr)
arr = arr+5
arr += 5 # added 5 to each element
print("ADD =",arr)
arr = arr*2
print("Multiply =",arr)
arr = arr**2 # exponent
print("Power of two =",arr)

arr1 = np.arange(11,21)
print(arr1)

print("arr + arr1 =",arr + arr1) # length should be same
print("arr * arr1 =",arr*arr1)
```

```python
x = np.arange(6).reshape(2,3)
y = np.arange(3)
print("x = ", x)
print()
print("y = ", y)
print()
print(x+y)
```

```python
# comparison operators - boolean arrays
x = np.arange(5)
y = np.arange(1,6)

print(x)
print(y)

print(x!=y) # checks all the elements
print(x>y)
print(x<y)
```

### The np.random Module

```python
# np.random.seed() + np.random.rand()
np.random.seed(3) # seed makes the internal formula static, value will be random but way of generation will be same
print(np.random.rand(4))  # range is from 0 to 1{ excluded }
print(np.random.randn(10)) # it creates random numbers in a normal distribution format
```

```python
# np.random.randit(low , high , size) low included , high excluded
print(np.random.randint(3,10,10))
```

```python
# np.random.choice() 
lst = [3,454,6,7,34,90,0]
np.random.choice(lst,6,replace=False) # replace = false - chooses form lst but do not repeat the element from lst
```

```python
#  np.random.shuffle()
z = np.arange(10)
np.random.shuffle(z) # suffles the whole array
print(z)
```

### Statistics

```python
#min ,max ,sum, mean, median, std
arr = np.arange(6).reshape(2,3)
print(arr)
print("min : ",arr.min())
print("row min : ",arr.min(axis=1))
print("col min : ",arr.min(axis=0))

print("sum of arr : ",arr.sum())
print("sum of row : ",arr.sum(axis=1))

print("mean of arr : ",np.mean(arr))
```
